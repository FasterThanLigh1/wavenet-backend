```python
# @title Chạy demo

from google.colab import drive
drive.mount('/content/drive')

import ipywidgets as widgets
from IPython.display import display
import numpy as np
import tensorflow as tf
from sklearn import preprocessing
from tensorflow.keras import layers, regularizers
from tensorflow.keras import backend as K
import joblib

# Thiết lập strategy
try:
    tpu = tf.distribute.cluster_resolver.TPUClusterResolver()
    tf.config.experimental_connect_to_cluster(tpu)
    tf.tpu.experimental.initialize_tpu_system(tpu)
    strategy = tf.distribute.experimental.TPUStrategy(tpu)
    print("Running on TPU:", tpu.master())
except ValueError:
    gpus = tf.config.list_physical_devices('GPU')
    if gpus:
        strategy = tf.distribute.get_strategy()
        print("Running on GPU")
    else:
        strategy = tf.distribute.get_strategy()
        print("Running on CPU")

# Định nghĩa hàm model_wavenet_timeseries (giữ nguyên như bạn đã có)
# Định nghĩa mô hình WaveNet cho dự báo đa mục tiêu
# len_seq: độ dài chuỗi đầu vào
def model_wavenet_timeseries(
    len_seq,
    len_out,
    dim_exo,
    dim_target,       # số kênh target (ví dụ 3)
    nb_filters,
    dim_filters,
    dilation_depth,
    use_bias,
    res_l2,
    final_l2,
    batch_size

):
    # Input exogenous series
    input_exo = tf.keras.Input(shape=(len_seq, dim_exo), name='input_exo')
    # Input multi-target series
    input_target = tf.keras.Input(shape=(len_seq, dim_target), name='input_target')

    outputs = []
    # Khởi tạo 2 biến sẽ cập nhật cho từng bước dự báo
    exo_inp = input_exo
    tgt_inp = input_target

    for t in range(len_out):
        # --- Causal Convolution cho exogenous ---
        pad_exo = layers.Lambda(
            lambda x: K.temporal_padding(x, padding=((dim_filters - 1), 0)),
            output_shape=lambda s: (s[0], s[1] + dim_filters - 1, s[2])
        )(exo_inp)
        conv_exo = layers.Conv1D(
            filters=nb_filters,
            kernel_size=dim_filters,
            dilation_rate=1,
            use_bias=use_bias,
            activation=None,
            kernel_regularizer=regularizers.l2(res_l2),
            name=f'causal_conv_exo_{t}'
        )(pad_exo)

        # --- Causal Convolution cho target ---
        pad_tgt = layers.Lambda(
            lambda x: K.temporal_padding(x, padding=((dim_filters - 1), 0)),
            output_shape=lambda s: (s[0], s[1] + dim_filters - 1, s[2])
        )(tgt_inp)
        conv_tgt = layers.Conv1D(
            filters=nb_filters,
            kernel_size=dim_filters,
            dilation_rate=1,
            use_bias=use_bias,
            activation=None,
            kernel_regularizer=regularizers.l2(res_l2),
            name=f'causal_conv_tgt_{t}'
        )(pad_tgt)

        skip_connections = []

        # --- Lớp dilated đầu tiên cho exogenous & target ---
        def dilated_block(x, filters_out, name_prefix):
            pad_amt = 2 * (dim_filters - 1)
            z = layers.Lambda(
                lambda y: K.temporal_padding(y, padding=(pad_amt, 0)),
                output_shape=lambda s: (s[0], s[1] + pad_amt, s[2])
            )(x)
            z = layers.Conv1D(
                filters=nb_filters,
                kernel_size=dim_filters,
                dilation_rate=2,
                activation='relu',
                use_bias=use_bias,
                kernel_regularizer=regularizers.l2(res_l2),
                name=f'{name_prefix}_dilated1'
            )(z)
            skip = layers.Conv1D(
                filters_out,
                kernel_size=1,
                padding='same',
                use_bias=False,
                kernel_regularizer=regularizers.l2(res_l2),
                name=f'{name_prefix}_skip1'
            )(z)
            res = layers.Conv1D(
                filters_out,
                kernel_size=1,
                padding='same',
                use_bias=False,
                kernel_regularizer=regularizers.l2(res_l2),
                name=f'{name_prefix}_res1'
            )(z)
            return skip, res

        skip_exo1, res_exo1 = dilated_block(conv_exo, dim_exo, f'exo_{t}')
        skip_tgt1, res_tgt1 = dilated_block(conv_tgt, dim_target, f'tgt_{t}')
        skip_connections += [layers.Concatenate(axis=-1)([skip_exo1, skip_tgt1])]

        # Cộng residual vào input
        exo_out = layers.Add()([exo_inp, res_exo1])
        tgt_out = layers.Add()([tgt_inp, res_tgt1])

        # Kết hợp exo & target
        out = layers.Concatenate(axis=-1)([exo_out, tgt_out])

        # --- Các lớp dilated tiếp theo ---
        for i in range(2, dilation_depth + 1):
            pad_amt = (2 ** i) * (dim_filters - 1)
            z = layers.Lambda(
                lambda y, p=pad_amt: K.temporal_padding(y, padding=(p, 0)),
                output_shape=lambda s, p=pad_amt: (s[0], s[1] + p, s[2])
            )(out)
            z = layers.Conv1D(
                filters=nb_filters,
                kernel_size=dim_filters,
                dilation_rate=2 ** i,
                activation='relu',
                use_bias=use_bias,
                kernel_regularizer=regularizers.l2(res_l2),
                name=f'dilated_conv_{i}_{t}'
            )(z)
            skip_i = layers.Conv1D(
                dim_exo + dim_target,
                kernel_size=1,
                padding='same',
                use_bias=False,
                kernel_regularizer=regularizers.l2(res_l2),
                name=f'skip_{i}_{t}'
            )(z)
            res_i = layers.Conv1D(
                dim_exo + dim_target,
                kernel_size=1,
                padding='same',
                use_bias=False,
                kernel_regularizer=regularizers.l2(res_l2),
                name=f'res_{i}_{t}'
            )(z)
            out = layers.Add()([out, res_i])
            skip_connections.append(skip_i)

        # Tổng hợp skip connections
        total_skip = layers.Add(name=f'total_skip_{t}')(skip_connections)
        # Output linear
        lin = layers.Activation('linear', name=f'lin_out_{t}')(total_skip)

        # Tạo dự báo multi-target cho bước t
        out_f_tgt = layers.Conv1D(
            dim_target,
            kernel_size=1,
            padding='same',
            kernel_regularizer=regularizers.l2(final_l2),
            name=f'out_f_tgt_{t}'
        )(lin)
        # Lấy step prediction
        step_pred = layers.Lambda(
            lambda y: y[:, -1:, :],
            output_shape=lambda s: (s[0], 1, dim_target)
        )(out_f_tgt)
        outputs.append(step_pred)

        # Cập nhật input window cho bước tiếp theo
        # Exogenous
        new_exo = layers.Concatenate(axis=1)([
            exo_inp,
            layers.Lambda(lambda y: y[:, -1:, :])(layers.Conv1D(dim_exo,1,padding='same')(lin))
        ])
        exo_inp = layers.Lambda(lambda y: y[:, 1:, :])(new_exo)
        # Target
        new_tgt = layers.Concatenate(axis=1)([
            tgt_inp,
            step_pred
        ])
        tgt_inp = layers.Lambda(lambda y: y[:, 1:, :])(new_tgt)

    # Stack outputs thành (batch, len_out, dim_target)
    stacked = layers.Lambda(
        lambda xs: tf.concat(xs, axis=1),
        output_shape=lambda s: (None, len_out, dim_target)
    )(outputs)

    model = tf.keras.Model([input_exo, input_target], stacked)
    return model

def compute_len_seq(dilation_depth):
    return (2 ** dilation_depth * 2)

# @title Tạo mô hình

def compute_len_seq(dilation_depth):
  return (2 ** dilation_depth * 2)

nb_filters = 96
dim_filters = 2
dilation_depth = 4     # len_seq = 32
use_bias = True
res_l2 = 0
final_l2 = 0

batch_size = 128
len_out = 48
dim_exo = 6
dim_target = 3

len_seq = compute_len_seq(dilation_depth)
print(len_seq)

# Tạo và tải mô hình trong scope
with strategy.scope():

    model = model_wavenet_timeseries(len_seq, len_out, dim_exo, dim_target, nb_filters, dim_filters, dilation_depth, use_bias, res_l2, final_l2, batch_size)
    # Đường dẫn đến file checkpoint đã lưu từ quá trình huấn luyện
    checkpoint_dir = "/content/drive/My Drive/Code-1/Thực tập 1-Dự báo phụ tải/Đề tài/checkpoints"
    checkpoint_path = os.path.join(checkpoint_dir, "model_checkpoint.keras")

    # Tải mô hình đã lưu (bao gồm cấu trúc, trọng số và optimizer)
    model = tf.keras.models.load_model(checkpoint_path, safe_mode=False)
# Tải scaler
scaler = joblib.load('/content/drive/My Drive/Code-1/Thực tập 1-Dự báo phụ tải/Đề tài/scaler.pkl')

# Tạo widget đầu vào
exo_input = widgets.Textarea(
    value='',
    placeholder='Nhập 32 giá trị nhiệt độ. Ví dụ:\n[14.53, 14.42, ..., 9.29]',
    description='Nhiệt độ:',
    layout={'width': '500px', 'height': '100px'}
)

target_input = widgets.Textarea(
    value='',
    placeholder='Nhập 32 giá trị mục tiêu của 3 khu vực. Ví dụ:\n[28641.8, 28130.8, ..., 36623.57]',
    description='Mục tiêu:',
    layout={'width': '500px', 'height': '100px'}
)

button = widgets.Button(description="Dự báo")
output = widgets.Output()

def on_button_clicked(b):
    with output:
        output.clear_output()
        try:
            # Bước 1: Đọc chuỗi exogenous và target từ widget
            exo_seq = np.array(eval(exo_input.value), dtype=np.float32)    # shape (len_seq,)
            target_seq = np.array(eval(target_input.value), dtype=np.float32)  # shape (len_seq,)

            # Đảm bảo đúng shape (len_seq, dim_exo) và (len_seq, dim_target)
            # Ở đây dim_exo=1, dim_target=1 (như ví dụ)
            exo_seq   = exo_seq.reshape(len_seq, dim_exo)       # (32,6) nếu dim_exo=6
            target_seq= target_seq.reshape(len_seq, dim_target) # (32,3) nếu dim_target=3

            # Bước 2: Chuẩn hóa cả hai chuỗi cùng lúc
            # Kết hợp để scaler.transform xử lý cùng scale
            combined = np.hstack([exo_seq, target_seq])  # (len_seq, dim_exo+dim_target)
            combined_norm = scaler.transform(combined)

            # Tách lại exo và target đã chuẩn hóa
            exo_norm    = combined_norm[:, :dim_exo]
            target_norm = combined_norm[:, dim_exo:]

            # Bước 3: Tạo batch input cho model
            x1 = exo_norm[np.newaxis, :, :].astype(np.float32)    # (1, len_seq, dim_exo)
            x2 = target_norm[np.newaxis, :, :].astype(np.float32) # (1, len_seq, dim_target)

            # In ra shape để kiểm tra
            print("x1 shape:", x1.shape)
            print("x2 shape:", x2.shape)

            # Bước 4: Dự báo
            pred_norm = model.predict([x1, x2])  # shape (1, len_out, dim_target)

            # Bước 5: Inverse transform riêng target
            # Để inverse chỉ target, cần đệm lại dummy exo bằng giá trị cuối
            inv_preds = []
            for i in range(len(pred_norm[0])):
                row_norm = np.zeros((1, dim_exo + dim_target), dtype=np.float32)
                # dùng exo_norm cuối cùng đưa vào phần exo
                row_norm[0, :dim_exo] = exo_norm[-1]
                # gán giá trị dự báo vào phần target
                row_norm[0, dim_exo:] = pred_norm[0, i]
                inv_row = scaler.inverse_transform(row_norm)
                inv_preds.append(inv_row[0, dim_exo:])

            inv_preds = np.vstack(inv_preds)  # shape (len_out, dim_target)

            # Bước 6: Hiển thị kết quả
            for step, vals in enumerate(inv_preds, 1):
                print(f"Bước {step}: {vals}")

        except Exception as e:
            print("Lỗi:", e)

button.on_click(on_button_clicked)
display(exo_input, area, target_input, button, output)
```

```python
# @title Chạy demo (Phiên bản hoàn chỉnh)

from google.colab import drive
drive.mount('/content/drive', force_remount=False)

import os
import numpy as np
import joblib
import ipywidgets as widgets
from IPython.display import display

import tensorflow as tf
from tensorflow.keras import layers, regularizers
from tensorflow.keras.layers import ZeroPadding1D, Concatenate
from tensorflow.keras import backend as K

# ================== THIẾT LẬP TPU/GPU ==================
try:
    tpu = tf.distribute.cluster_resolver.TPUClusterResolver()
    tf.config.experimental_connect_to_cluster(tpu)
    tf.tpu.experimental.initialize_tpu_system(tpu)
    strategy = tf.distribute.experimental.TPUStrategy(tpu)
    print("Running on TPU      :", tpu.master())
except ValueError:
    gpus = tf.config.list_physical_devices('GPU')
    if gpus:
        strategy = tf.distribute.get_strategy()
        print("Running on GPU")
    else:
        strategy = tf.distribute.get_strategy()
        print("Running on CPU")

# ================== MODEL DEFINITION ==================
def model_wavenet_timeseries(
    len_seq, len_out, dim_exo, dim_tgt,
    nb_filters, dim_filters, dilation_depth,
    use_bias, res_l2, final_l2
):
    inp_exo = tf.keras.Input((len_seq, dim_exo), name='input_exo')
    inp_tgt = tf.keras.Input((len_seq, dim_tgt), name='input_target')

    exo, tgt = inp_exo, inp_tgt
    outputs = []

    for t in range(len_out):
        # Causal conv exo
        p = dim_filters-1
        pad_exo = ZeroPadding1D((p,0), name=f'pad_exo_{t}')(exo)
        conv_exo = layers.Conv1D(nb_filters, dim_filters, dilation_rate=1,
                                 use_bias=use_bias, activation=None,
                                 kernel_regularizer=regularizers.l2(res_l2),
                                 name=f'causal_conv_exo_{t}')(pad_exo)

        # Causal conv tgt
        pad_tgt = ZeroPadding1D((p,0), name=f'pad_tgt_{t}')(tgt)
        conv_tgt = layers.Conv1D(nb_filters, dim_filters, dilation_rate=1,
                                 use_bias=use_bias, activation=None,
                                 kernel_regularizer=regularizers.l2(res_l2),
                                 name=f'causal_conv_tgt_{t}')(pad_tgt)

        skips = []
        # first dilated block
        def dilated_block(x, out_ch, name):
            pad_amt = 2*(dim_filters-1)
            z = ZeroPadding1D((pad_amt,0), name=f'{name}_pad')(x)
            z = layers.Conv1D(nb_filters, dim_filters, dilation_rate=2,
                              activation='relu', use_bias=use_bias,
                              kernel_regularizer=regularizers.l2(res_l2),
                              name=f'{name}_dilated')(z)
            skip = layers.Conv1D(out_ch, 1, padding='same', use_bias=False,
                                 kernel_regularizer=regularizers.l2(res_l2),
                                 name=f'{name}_skip')(z)
            res  = layers.Conv1D(out_ch, 1, padding='same', use_bias=False,
                                 kernel_regularizer=regularizers.l2(res_l2),
                                 name=f'{name}_res')(z)
            return skip, res

        skip_e, res_e = dilated_block(conv_exo, dim_exo, f'exo_{t}')
        skip_t, res_t = dilated_block(conv_tgt, dim_tgt, f'tgt_{t}')
        skips.append(Concatenate(name=f'skip_cat_{t}')([skip_e, skip_t]))

        # residual add
        exo = layers.Add(name=f'exo_resadd_{t}')([exo, res_e])
        tgt = layers.Add(name=f'tgt_resadd_{t}')([tgt, res_t])

        x = Concatenate(name=f'exo_tgt_cat_{t}')([exo, tgt])

        # further dilated blocks
        for i in range(2, dilation_depth+1):
            pad_amt = (2**i)*(dim_filters-1)
            z = ZeroPadding1D((pad_amt,0), name=f'pad_{i}_{t}')(x)
            z = layers.Conv1D(nb_filters, dim_filters, dilation_rate=2**i,
                              activation='relu', use_bias=use_bias,
                              kernel_regularizer=regularizers.l2(res_l2),
                              name=f'dilated_{i}_{t}')(z)
            skip_i = layers.Conv1D(dim_exo+dim_tgt, 1, padding='same',
                                   use_bias=False,
                                   kernel_regularizer=regularizers.l2(res_l2),
                                   name=f'skip_{i}_{t}')(z)
            res_i  = layers.Conv1D(dim_exo+dim_tgt, 1, padding='same',
                                   use_bias=False,
                                   kernel_regularizer=regularizers.l2(res_l2),
                                   name=f'res_{i}_{t}')(z)
            x = layers.Add(name=f'resadd_{i}_{t}')([x, res_i])
            skips.append(skip_i)

        total = layers.Add(name=f'total_skip_{t}')(skips)
        lin   = layers.Activation('linear', name=f'lin_{t}')(total)

        out_t = layers.Conv1D(dim_tgt, 1, padding='same',
                              kernel_regularizer=regularizers.l2(final_l2),
                              name=f'out_tgt_{t}')(lin)
        step = layers.Lambda(lambda y: y[:, -1:, :],
                             name=f'step_{t}')(out_t)
        outputs.append(step)

        # update windows
        exo = Concatenate(axis=1, name=f'upd_exo_{t}')([
            exo, layers.Conv1D(dim_exo,1,padding='same',
                               name=f'upd_exo_conv_{t}')(lin)[:, -1:, :]
        ])
        exo = layers.Lambda(lambda y: y[:,1:,:], name=f'upd_exo_slice_{t}')(exo)

        tgt = Concatenate(axis=1, name=f'upd_tgt_{t}')([tgt, step])
        tgt = layers.Lambda(lambda y: y[:,1:,:], name=f'upd_tgt_slice_{t}')(tgt)

    y_out = layers.Lambda(lambda ls: tf.concat(ls,axis=1),
                          name='y_out')(outputs)

    return tf.keras.Model([inp_exo, inp_tgt], y_out)


# ================== LOAD MODEL & SCALER ==================
nb_filters    = 96
dim_filters   = 2
dilation_depth= 4
use_bias      = True
res_l2        = 0
final_l2      = 0

batch_size    = 128
len_out       = 48
dim_exo       = 6
dim_target    = 3
len_seq       = (2**dilation_depth)*2

checkpoint_dir  = "/content/drive/My Drive/Code-1/Thực tập 1-Dự báo phụ tải/Đề tài/checkpoints"
checkpoint_path = os.path.join(checkpoint_dir, "model_checkpoint.keras")

scaler_path   = '/content/drive/My Drive/Code-1/Thực tập 1-Dự báo phụ tải/Đề tài/scaler.pkl'

with strategy.scope():
    model = model_wavenet_timeseries(
        len_seq, len_out, dim_exo, dim_target,
        nb_filters, dim_filters, dilation_depth,
        use_bias, res_l2, final_l2
    )
    model.load_weights(checkpoint_path)

scaler = joblib.load(scaler_path)

# ================== WIDGETS ==================
exo_input = widgets.Textarea(
    placeholder='Nhập {} giá trị Exo, dạng: [..]'.format(len_seq),
    description='Exo:', layout={'width':'500px','height':'80px'}
)
target_input = widgets.Textarea(
    placeholder='Nhập {} giá trị Target, dạng: [..]'.format(len_seq),
    description='Tgt:', layout={'width':'500px','height':'80px'}
)
button = widgets.Button(description="Dự báo")
output = widgets.Output()

def on_button_clicked(b):
    with output:
        output.clear_output()
        try:
            # đọc và reshape
            exo = np.array(eval(exo_input.value),dtype=np.float32).reshape(len_seq,dim_exo)
            tgt = np.array(eval(target_input.value),dtype=np.float32).reshape(len_seq,dim_target)

            # normalize chung
            comb = np.hstack([exo, tgt])
            norm = scaler.transform(comb)
            exo_n = norm[:,:dim_exo][None,...]
            tgt_n = norm[:,dim_exo:][None,...]

            # predict
            pred_n = model.predict([exo_n, tgt_n])[0]  # (len_out, dim_target)

            # inverse và hiển thị
            for i, row in enumerate(pred_n,1):
                buf = np.zeros((1, dim_exo+dim_target),dtype=np.float32)
                buf[0,:dim_exo] = exo[-1]
                buf[0,dim_exo:] = row
                inv = scaler.inverse_transform(buf)[0,dim_exo:]
                print(f"Bước {i}: {inv}")

        except Exception as e:
            print("Lỗi:", e)

button.on_click(on_button_clicked)
display(exo_input, target_input, button, output)
```

```python
# @title Chạy demo 2

from google.colab import drive
drive.mount('/content/drive')

import ipywidgets as widgets
from IPython.display import display, clear_output
import numpy as np
import tensorflow as tf
import joblib
from tensorflow.keras.layers import Lambda
from tensorflow.keras import backend as K

# Khởi tạo mô hình theo chiến lược đã có (TPU/GPU/CPU)
try:
    tpu = tf.distribute.cluster_resolver.TPUClusterResolver()
    tf.config.experimental_connect_to_cluster(tpu)
    tf.tpu.experimental.initialize_tpu_system(tpu)
    strategy = tf.distribute.experimental.TPUStrategy(tpu)
    print("Running on TPU:", tpu.master())
except ValueError:
    gpus = tf.config.list_physical_devices('GPU')
    if gpus:
        strategy = tf.distribute.get_strategy()
        print("Running on GPU")
    else:
        strategy = tf.distribute.get_strategy()
        print("Running on CPU")

def model_wavenet_timeseries(len_seq, len_out, dim_exo, nb_filters, dim_filters,
                             dilation_depth, use_bias, res_l2, final_l2, batch_size):
    input_shape_exo = tf.keras.Input(shape=(len_seq, dim_exo), name='input_exo')
    input_shape_target = tf.keras.Input(shape=(len_seq, 1), name='input_target')

    input_exo = input_shape_exo
    input_target = input_shape_target

    outputs = []
    for t in range(len_out):
        out_exo = Lambda(lambda x: K.temporal_padding(x, padding=((dim_filters - 1), 0)),
                         output_shape=lambda s: (s[0], s[1] + (dim_filters - 1), s[2]))(input_exo)
        out_exo = tf.keras.layers.Conv1D(filters=nb_filters, kernel_size=dim_filters, dilation_rate=1,
                                         use_bias=use_bias, activation=None,
                                         name='causal_convolution_exo_%i' % t,
                                         kernel_regularizer=tf.keras.regularizers.l2(res_l2))(out_exo)

        out_target = Lambda(lambda x: K.temporal_padding(x, padding=((dim_filters - 1), 0)),
                            output_shape=lambda s: (s[0], s[1] + (dim_filters - 1), s[2]))(input_target)
        out_target = tf.keras.layers.Conv1D(filters=nb_filters, kernel_size=dim_filters, dilation_rate=1,
                                            use_bias=use_bias, activation=None,
                                            name='causal_convolution_target_%i' % t,
                                            kernel_regularizer=tf.keras.regularizers.l2(res_l2))(out_target)

        skip_outputs = []
        z_exo = Lambda(lambda x: K.temporal_padding(x, padding=(2*(dim_filters - 1), 0)),
                       output_shape=lambda s: (s[0], s[1] + 2*(dim_filters - 1), s[2]))(out_exo)
        z_exo = tf.keras.layers.Conv1D(filters=nb_filters, kernel_size=dim_filters, dilation_rate=2,
                                       use_bias=use_bias, activation="relu",
                                       name='dilated_convolution_exo_1_%i' % t,
                                       kernel_regularizer=tf.keras.regularizers.l2(res_l2))(z_exo)
        skip_exo = tf.keras.layers.Conv1D(dim_exo, 1, padding='same', use_bias=False,
                                          kernel_regularizer=tf.keras.regularizers.l2(res_l2))(z_exo)
        first_out_exo = tf.keras.layers.Conv1D(dim_exo, 1, padding='same', use_bias=False,
                                               kernel_regularizer=tf.keras.regularizers.l2(res_l2))(z_exo)

        z_target = Lambda(lambda x: K.temporal_padding(x, padding=(2*(dim_filters - 1), 0)),
                          output_shape=lambda s: (s[0], s[1] + 2*(dim_filters - 1), s[2]))(out_target)
        z_target = tf.keras.layers.Conv1D(filters=nb_filters, kernel_size=dim_filters, dilation_rate=2,
                                          use_bias=use_bias, activation="relu",
                                          name='dilated_convolution_target_1_%i' % t,
                                          kernel_regularizer=tf.keras.regularizers.l2(res_l2))(z_target)
        skip_target = tf.keras.layers.Conv1D(1, 1, padding='same', use_bias=False,
                                               kernel_regularizer=tf.keras.regularizers.l2(res_l2))(z_target)
        first_out_target = tf.keras.layers.Conv1D(1, 1, padding='same', use_bias=False,
                                                    kernel_regularizer=tf.keras.regularizers.l2(res_l2))(z_target)

        concat_skip = Lambda(lambda x: tf.concat(x, axis=2),
                             output_shape=lambda s: (s[0][0], s[0][1], s[0][2] + s[1][2]))([skip_exo, skip_target])
        skip_outputs.append(concat_skip)

        res_exo_out = tf.keras.layers.Add()([input_exo, first_out_exo])
        res_target_out = tf.keras.layers.Add()([input_target, first_out_target])
        out_concat = Lambda(lambda x: tf.concat(x, axis=2),
                            output_shape=lambda s: (s[0][0], s[0][1], s[0][2] + s[1][2]))([res_exo_out, res_target_out])
        out = out_concat

        for i in range(2, dilation_depth + 1):
            pad_amount = 2**i * (dim_filters - 1)
            z = Lambda(lambda x, pad=pad_amount: K.temporal_padding(x, padding=(pad, 0)),
                       output_shape=lambda s, pad=pad_amount: (s[0], s[1] + pad, s[2]))(out_concat)
            z = tf.keras.layers.Conv1D(filters=nb_filters, kernel_size=dim_filters,
                                       dilation_rate=2**i, use_bias=use_bias, activation="relu",
                                       name='dilated_convolution_%i_%i' % (i, t),
                                       kernel_regularizer=tf.keras.regularizers.l2(res_l2))(z)

            skip_x = tf.keras.layers.Conv1D(dim_exo+1, 1, padding='same', use_bias=False,
                                            kernel_regularizer=tf.keras.regularizers.l2(res_l2))(z)
            first_out_x = tf.keras.layers.Conv1D(dim_exo+1, 1, padding='same', use_bias=False,
                                                 kernel_regularizer=tf.keras.regularizers.l2(res_l2))(z)
            res_x = tf.keras.layers.Add()([out, first_out_x])
            out = res_x
            skip_outputs.append(skip_x)

        out = tf.keras.layers.Add()(skip_outputs)
        out = tf.keras.layers.Activation('linear', name="output_linear_%i" % t)(out)
        out_f_target = tf.keras.layers.Conv1D(1, 1, padding='same',
                                                kernel_regularizer=tf.keras.regularizers.l2(final_l2))(out)
        out_f_exo = tf.keras.layers.Conv1D(dim_exo, 1, padding='same',
                                             kernel_regularizer=tf.keras.regularizers.l2(final_l2))(out)
        outputs.append(Lambda(lambda x: x[:, -1:, :],
                              output_shape=lambda s: (s[0], 1, s[2]))(out_f_target))

        new_exo = Lambda(lambda x: tf.concat(x, axis=1),
                         output_shape=lambda s: (s[0][0], s[0][1] + s[1][1], s[0][2]))([
                             input_exo,
                             Lambda(lambda x: x[:, -1:, :],
                                    output_shape=lambda s: (s[0], 1, s[2]))(out_f_exo)
                         ])
        input_exo = Lambda(lambda x: x[:, 1:, :],
                           output_shape=lambda s: (s[0], s[1]-1, s[2]))(new_exo)

        new_target = Lambda(lambda x: tf.concat(x, axis=1),
                            output_shape=lambda s: (s[0][0], s[0][1] + s[1][1], s[0][2]))([
                                input_target,
                                Lambda(lambda x: x[:, -1:, :],
                                       output_shape=lambda s: (s[0], 1, s[2]))(out_f_target)
                            ])
        input_target = Lambda(lambda x: x[:, 1:, :],
                              output_shape=lambda s: (s[0], s[1]-1, s[2]))(new_target)

    outputs = Lambda(lambda x: tf.transpose(tf.squeeze(tf.stack(x), axis=-1), perm=[1, 0, 2]),
                     output_shape=lambda s: (None, len_out, 1))(outputs)
    model = tf.keras.Model([input_shape_exo, input_shape_target], outputs)
    return model

def compute_len_seq(dilation_depth):
    return (2 ** dilation_depth * 2)

# Các tham số
nb_filters = 96
dim_filters = 2
dilation_depth = 4
use_bias = True
res_l2 = 0
final_l2 = 0
batch_size = 128
len_out = 18
dim_exo = 4
len_seq = compute_len_seq(dilation_depth)  # len_seq = 32

with strategy.scope():
    model = model_wavenet_timeseries(len_seq, len_out, dim_exo, nb_filters, dim_filters,
                                     dilation_depth, use_bias, res_l2, final_l2, batch_size=batch_size)
    weights_path = "/content/drive/My Drive/Code-1/Thực tập 1-Dự báo phụ tải/weights_train_2.hdf5"
    model.load_weights(weights_path)

scaler = joblib.load('/content/drive/My Drive/Code-1/Thực tập 1-Dự báo phụ tải/scaler.pkl')


# ------------------ Giao diện nhập liệu ------------------
# Tạo widget đầu vào
exo_input = widgets.Textarea(
    value='',
    placeholder='Nhập 32 giá trị nhiệt độ. Ví dụ:\n[14.53, 14.42, ..., 9.29]',
    description='Nhiệt độ:',
    layout={'width': '500px', 'height': '100px'}
)

area = widgets.Dropdown(
    options=['1', '2', '3'],
    value='1',
    description='Khu vực:'
)

target_input = widgets.Textarea(
    value='',
    placeholder='Nhập 32 giá trị mục tiêu. Ví dụ:\n[28641.8, 28130.8, ..., 36623.57]',
    description='Mục tiêu:',
    layout={'width': '500px', 'height': '100px'}
)

button = widgets.Button(description="Dự báo")
output = widgets.Output()

def on_button_clicked(b):
    with output:
        clear_output(wait=True)
        try:
            # Bước 1: Đọc input
            exo_seq = np.array(eval(exo_input.value)).reshape(32, 1)  # (32,1)
            area_input = area.value.strip()
            if area_input not in ['1', '2', '3']:
                raise ValueError("Giá trị khu vực không hợp lệ. Nhập 1, 2 hoặc 3.")
            target_seq = np.array(eval(target_input.value)).reshape(32, 1)  # (32,1)

            # Bước 2: Tạo mảng đầy đủ với các cột giả
            if area.value == '1':
                dummy_zone2 = np.zeros((32, 1))  # Cột giả cho Zone2
                dummy_zone3 = np.zeros((32, 1))  # Cột giả cho Zone3
                input_seq = np.hstack((exo_seq, target_seq, dummy_zone2, dummy_zone3))  # Shape: (32, 4)
            elif area.value == '2':
                dummy_zone1 = np.zeros((32, 1))  # Cột giả cho Zone1
                dummy_zone3 = np.zeros((32, 1))  # Cột giả cho Zone3
                input_seq = np.hstack((exo_seq, dummy_zone1, target_seq, dummy_zone3))  # Shape: (32, 4)
            elif area.value == '3':
                dummy_zone1 = np.zeros((32, 1))  # Cột giả cho Zone1
                dummy_zone2 = np.zeros((32, 1))  # Cột giả cho Zone2
                input_seq = np.hstack((exo_seq, dummy_zone1, dummy_zone2, target_seq))  # Shape: (32, 4)
            else:
                raise ValueError("Invalid area value. Please enter 1, 2, or 3.")

            # Bước 3: Chuẩn hóa dữ liệu
            input_seq_norm = scaler.transform(input_seq)  # Shape: (32, 4)
            exo_norm = input_seq_norm[:, :1]  # cột nhiệt độ

            # Tùy thuộc vào khu vực, tách cột phụ tải (PowerConsumption) từ input_seq_norm và tạo one-hot tương ứng
            if area.value == '1':
                target_norm = input_seq_norm[:, 1:2]  # cột phụ tải của Zone1
                one_hot = np.tile(np.array([1, 0, 0]), (32, 1))  # zone 1 → [1,0,0]
            elif area.value == '2':
                target_norm = input_seq_norm[:, 2:3]  # cột phụ tải của Zone2
                one_hot = np.tile(np.array([0, 1, 0]), (32, 1))  # zone 2 → [0,1,0]
            elif area.value == '3':
                target_norm = input_seq_norm[:, 3:4]  # cột phụ tải của Zone3
                one_hot = np.tile(np.array([0, 0, 1]), (32, 1))  # zone 3 → [0,0,1]
            else:
                raise ValueError("Invalid area value. Please enter 1, 2, or 3.")

            transformed_input = np.hstack((exo_norm, one_hot, target_norm))
            print("Final input shape:", transformed_input.shape)

            # Bước 4: Tách dữ liệu thành seriesx và seriesy
            seriesx = transformed_input[:, :-1]   # (32, 4)
            seriesy = transformed_input[:, -1:]     # (32, 1)
            x1 = seriesx[np.newaxis, :, :].astype(np.float32)  # (1, 32, 4)
            x2 = seriesy[np.newaxis, :, :].astype(np.float32)    # (1, 32, 1)
            print("x1 shape:", x1.shape)
            print("x2 shape:", x2.shape)

            # Bước 5: Dự báo
            prediction_norm = model.predict([x1, x2])
            pred_values_norm = prediction_norm[0, :, 0]  # (18,)

            # Tính giá trị trung tính cho mỗi cột dựa trên MinMaxScaler (MinMaxScaler tính toán riêng từng cột)
            neutral = ((scaler.data_min_ + scaler.data_max_) / 2).reshape(1, -1)  # shape: (1, 4)

            # Bước 6: In ra kết quả sau khi inverse transform
            predictions = []
            for pred in pred_values_norm:
                # Sử dụng copy của neutral làm dummy_output
                dummy_output = neutral.copy()
                # Cột 0: sử dụng giá trị nhiệt độ cuối cùng (chưa chuẩn hóa) từ exo_seq
                dummy_output[0, 0] = exo_seq[-1, 0]

                # Gán giá trị dự báo đã chuẩn hóa vào cột phụ tải tương ứng với khu vực dự báo
                if area_input == '1':
                    dummy_output[0, 1] = pred
                elif area_input == '2':
                    dummy_output[0, 2] = pred
                elif area_input == '3':
                    dummy_output[0, 3] = pred

                # Áp dụng inverse_transform để phục hồi giá trị gốc
                inv = scaler.inverse_transform(dummy_output)

                # Lấy giá trị phục hồi của cột phụ tải tương ứng với khu vực dự báo
                if area_input == '1':
                    predictions.append(inv[0, 1])
                elif area_input == '2':
                    predictions.append(inv[0, 2])
                elif area_input == '3':
                    predictions.append(inv[0, 3])

            # In ra kết quả dự báo
            for i, p in enumerate(predictions):
                print(f"Giá trị dự báo bước {i+1}: {p:.2f}")


        except Exception as e:
            print("Lỗi:", str(e))

button.on_click(on_button_clicked)

# Hiển thị giao diện: chỉ cần chạy cell này 1 lần, sau đó mỗi lần thay đổi input và nhấn button sẽ cập nhật kết quả.
display(exo_input, area, target_input, button, output)
```
